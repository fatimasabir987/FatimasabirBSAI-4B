Visualizing image in different color spaces
RGB Image :

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")

plt.imshow(img)
```

Gray Scale Image :

```python
import cv2

img = cv2.imread("D:/open cv/Image Processing and Enhancement/g4g.png", 0)

if img is None:
    print("Image not found!")
else:
    cv2.imshow("Image", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

YCrCb Color Space :

```python
import cv2

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")

img = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)

cv2.imshow('image', img) 

cv2.waitKey(0)         
cv2.destroyAllWindows()
```

HSV color space :

```python
import cv2

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")


cv2.imshow('image', img) 

cv2.waitKey(0)         
cv2.destroyAllWindows()
```

LAB color space :

```python
import cv2

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")

img = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

cv2.imshow('image', img) 

cv2.waitKey(0)         
cv2.destroyAllWindows()
```

Edge map of image :

```python
import cv2

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")


laplacian = cv2.Laplacian(img, cv2.CV_64F)
cv2.imshow('EdgeMap', laplacian) 

cv2.waitKey(0)         
cv2.destroyAllWindows()
```

Heat map of image :

```python
import matplotlib.pyplot as plt
import cv2

img = cv2.imread(""D:/open cv/Feature Detection and Description/babarazam.jpeg"")

plt.imshow(img, cmap ='hot')
```

Spectral Image map :

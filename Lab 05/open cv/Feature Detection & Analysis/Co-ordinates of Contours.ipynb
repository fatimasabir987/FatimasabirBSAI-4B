```python
import cv2
img_path = "D:/open cv/Feature Detection and Description/babarazam.jpeg"  
img2 = cv2.imread(img_path, cv2.IMREAD_COLOR)
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

if img2 is None or img is None:
    raise ValueError(f"Image not found at {img_path}")

import numpy as np

font = cv2.FONT_HERSHEY_COMPLEX

_, threshold = cv2.threshold(img, 110, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(threshold, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

for cnt in contours:
    approx = cv2.approxPolyDP(cnt, 0.009 * cv2.arcLength(cnt, True), True)
    cv2.drawContours(img2, [approx], 0, (0, 0, 255), 5)

    n = approx.ravel()
    for i in range(0, len(n), 2):
        x, y = n[i], n[i + 1]
        coord = f"{x} {y}"
        if i == 0:
            cv2.putText(img2, "Arrow tip", (x, y), font, 0.5, (255, 0, 0))
        else:
            cv2.putText(img2, coord, (x, y), font, 0.5, (0, 255, 0))

cv2.imshow('Contours with Coordinates', img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

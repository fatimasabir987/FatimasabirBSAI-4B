```python
import numpy as np
import cv2

cap = cv2.VideoCapture('D:/open cv/Advanced Image Manipulation & Background Subtraction/15297904_3840_2160_24fps.mp4')

fgbg = cv2.createBackgroundSubtractorMOG2()

while True:
    ret, frame = cap.read()
    if not ret:
        break  
    fgmask = fgbg.apply(frame)

    cv2.imshow('Original Frame', frame)
    cv2.imshow('Foreground Mask', fgmask)

    if cv2.waitKey(30) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

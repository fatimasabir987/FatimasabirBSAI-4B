```python
import cv2
import numpy as np

img1_color = cv2.imread("D:/open cv/Advanced Image Manipulation & Background Subtraction/im1-copy.webp")  
img2_color = cv2.imread("D:/open cv/Advanced Image Manipulation & Background Subtraction/im2-copy.webp")  
if img1_color is None or img2_color is None:
    print("Error: One or both images not found.")
    exit()

img1 = cv2.cvtColor(img1_color, cv2.COLOR_BGR2GRAY)
img2 = cv2.cvtColor(img2_color, cv2.COLOR_BGR2GRAY)
height, width = img2.shape

orb_detector = cv2.ORB_create(5000)

kp1, d1 = orb_detector.detectAndCompute(img1, None)
kp2, d2 = orb_detector.detectAndCompute(img2, None)

matcher = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches = matcher.match(d1, d2)
matches = sorted(matches, key=lambda x: x.distance)  

matches = matches[:int(len(matches)*0.9)]
no_of_matches = len(matches)
print(f"Number of good matches: {no_of_matches}")

p1 = np.zeros((no_of_matches, 2))
p2 = np.zeros((no_of_matches, 2))

for i in range(no_of_matches):
    p1[i, :] = kp1[matches[i].queryIdx].pt
    p2[i, :] = kp2[matches[i].trainIdx].pt

homography, mask = cv2.findHomography(p1, p2, cv2.RANSAC)
if homography is not None:
    print("Homography found: True")
else:
    print("Homography found: False")
    exit()

transformed_img = cv2.warpPerspective(img1_color, homography, (width, height))

cv2.imwrite('output.jpg', transformed_img)
print("Output image saved as 'output.jpg'")

cv2.imshow("Aligned Image", transformed_img)
cv2.imshow("Reference Image", img2_color)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
img1_color = cv2.imread("D:/open cv/Advanced Image Manipulation & Background Subtraction/im1-copy.webp") 
img2_color = cv2.imread("D:/open cv/Advanced Image Manipulation & Background Subtraction/im2-copy.webp") 

if img1_color is None or img2_color is None:
    print("Error: One or both images not found!")
    exit()
img1_gray = cv2.cvtColor(img1_color, cv2.COLOR_BGR2GRAY)
img2_gray = cv2.cvtColor(img2_color, cv2.COLOR_BGR2GRAY)
height, width = img2_gray.shape

orb = cv2.ORB_create(5000)
kp1, des1 = orb.detectAndCompute(img1_gray, None)
kp2, des2 = orb.detectAndCompute(img2_gray, None)

bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches = bf.match(des1, des2)
matches = sorted(matches, key=lambda x: x.distance)

matches = matches[:int(len(matches) * 0.9)]
print(f"Number of good matches used: {len(matches)}")

p1 = np.zeros((len(matches), 2))
p2 = np.zeros((len(matches), 2))

for i, match in enumerate(matches):
    p1[i, :] = kp1[match.queryIdx].pt
    p2[i, :] = kp2[match.trainIdx].pt

homography, mask = cv2.findHomography(p1, p2, cv2.RANSAC)
if homography is not None:
    print("Homography found: True")
else:
    print("Homography found: False")
    exit()

aligned_img = cv2.warpPerspective(img1_color, homography, (width, height))

match_img = cv2.drawMatches(img1_color, kp1, img2_color, kp2, matches[:20], None, flags=2)

success1 = cv2.imwrite('aligned_output.jpg', aligned_img)
success2 = cv2.imwrite('matches_visual.jpg', match_img)

print(f"Aligned image saved: {success1}")
print(f"Keypoints matches image saved: {success2}")

cv2.imshow("Aligned Image", aligned_img)
cv2.imshow("Top 20 Matches", match_img)
cv2.imshow("Reference Image", img2_color)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

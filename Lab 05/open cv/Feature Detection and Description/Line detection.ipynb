```python
import cv2
import numpy as np
img_path = "D:/open cv/Feature Detection and Description/line-detection-2.png"
img = cv2.imread(img_path)

if img is None:
    print("Error: Image not found! Check the file path.")
    exit()

print("Image loaded successfully.")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 50, 150, apertureSize=3)
print("Edge detection completed.")

lines = cv2.HoughLines(edges, 1, np.pi/180, 200)

if lines is not None:
    print(f"Number of lines detected: {len(lines)}")

    for r_theta in lines:
        r, theta = r_theta[0]

        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * r
        y0 = b * r

        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))

        cv2.line(img, (x1, y1), (x2, y2), (0, 0, 255), 2)
else:
    print("No lines detected!")

success = cv2.imwrite('linesDetected.jpg', img)
print("Lines detected image saved:", success)

cv2.imshow("Detected Lines", img)
print("Press 'q' to close the window.")

while True:
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img_path = 'D:/open cv/Feature Detection and Description/linesDetected.jpg'
image = cv2.imread(img_path)

if image is None:
    print("Error: Image not found! Check the path.")
    exit()
else:
    print("Image loaded successfully.")


gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 50, 150, apertureSize=3)
print("Edge detection completed.")


lines_list = []
lines = cv2.HoughLinesP(
    edges,
    1,             
    np.pi / 180,    
    threshold=100,  
    minLineLength=5,
    maxLineGap=10
)

if lines is not None:
    print(f"Number of lines detected: {len(lines)}")
    for points in lines:
        x1, y1, x2, y2 = points[0]
        cv2.line(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        lines_list.append([(x1, y1), (x2, y2)])
else:
    print("No lines detected!")

success = cv2.imwrite('detectedLines_final.png', image)
print("Detected lines image saved:", success)

cv2.imshow("Detected Lines", image)
print("Press 'q' to close the window.")

while True:
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
